# Matplotlib -- title and labels

```python
import matplotlib.pyplot as plt
import numpy as np 
x= np.array([30,40,50,60,70])
y = np.array([120,130,100,150,160])
plt.plot(x,y,'o-r', ms = 10,linewidth= '2',mfc = 'k')
plt.title("HEALTH MONITOR")
plt.xlabel("Average oxygen")
plt.ylabel("blood pressure")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np 
x= np.array([30,40,50,60,70])
y = np.array([120,130,100,150,160])
font1 = {'family': 'serif', 'color':'blue', 'size': 15}
font2 = {'family': "serif", 'color':'black', 'size': 10}
plt.plot(x,y)
plt.title("HEALTH MONITOR", fontdict = font1)
plt.xlabel("Average oxygen", fontdict = font2)
plt.ylabel("blood pressure", fontdict = font2)
plt.show()
```

# changing the location of title

```python
import matplotlib.pyplot as plt
import numpy as np 
x= np.array([30,40,50,60,70])
y = np.array([120,130,100,150,160])
font1 = {'family': 'serif', 'color':'blue', 'size': 15}
font2 = {'family': "serif", 'color':'black', 'size': 10}
plt.plot(x,y)
plt.title("HEALTH MONITOR", fontdict = font1, loc = 'left')#loc- location
plt.xlabel("Average oxygen", fontdict = font2)
plt.ylabel("blood pressure", fontdict = font2)
plt.show()
```

# gridlines.py via grid()

```python
# legal values are x and y and both default values is both
import matplotlib.pyplot as plt
import numpy as np 
x= np.array([30,40,50,60,70])
y = np.array([120,130,100,150,160])
font1 = {'family': 'serif', 'color':'blue', 'size': 15}
font2 = {'family': "serif", 'color':'black', 'size': 10}
plt.plot(x,y)
plt.title("HEALTH MONITOR", fontdict = font1)
plt.xlabel("Average oxygen", fontdict = font2)
plt.ylabel("blood pressure", fontdict = font2)
plt.grid()
# we can specify which grid lines to display via x or y axis
plt.grid(axis='x')
plt.show()
```

```python
# setting up tje line properties from grid
import matplotlib.pyplot as plt
import numpy as np 
x= np.array([30,40,50,60,70])
y = np.array([120,130,100,150,160])
font1 = {'family': 'serif', 'color':'blue', 'size': 15}
font2 = {'family': "serif", 'color':'black', 'size': 10}

plt.title("HEALTH MONITOR", fontdict = font1)
plt.xlabel("Average oxygen", fontdict = font2)
plt.ylabel("blood pressure", fontdict = font2)
plt.plot(x,y)
plt.grid(color = 'green', linestyle ='--', linewidth =1)
plt.show()
```

# subplot

```python
# display the multiple plots- with subplot() 
# you can draw multiple plots in one figure
import matplotlib.pyplot as plt
import numpy as np 
# plot1
x = np.array([0,1,2,3])
y = np.array([3,8,1,10])
plt.subplot(1,2,1)#the figure has 1 row, 2 coloumn and this plot is plot1
plt.plot(x,y)
# plot2
x = np.array([1,1,3,2])
y = np.array([10,20,10,30])
plt.subplot(1,2,2)
plt.plot(x,y)
plt.show()
```

# Alpha 

```python
# alpha- to adjust the transperency of the dots
import matplotlib.pyplot as plt
import numpy as np


x = np.array([11,12,40,50,34,98,76,87,39])
y = np.array([1,5,7,9,3,4,5,8,9])
colors= ([0,10,30,35,50,60,100,80,95])
sizes =np.array([20,30,40,50,600,800,1000,1200,1500])
plt.scatter(x,y, c= colors, cmap="viridis", s = sizes, alpha= 0.5)

plt.colorbar()
plt.show()
```

# Combining color, size and alpha 

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.random.randint(100, size = (100))
y = np.random.randint(100, size = (100))
colors = np.random.randint(100, size = (100))
sizes = 10*np.random.randint(100, size = (100))
plt.scatter(x,y, c= colors, cmap="nipy_spectral", s = sizes, alpha= 0.5)

plt.colorbar()
plt.show()
```
