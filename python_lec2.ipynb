```python
# return kab use karte hain:
# Jab function ka result aage kahin use karna ho — kisi variable me store karna ho, kisi aur function me pass karna ho, ya calculation aage badhani ho.

# print kab use karte hain:
# Jab bas result screen pe dikhana ho, use aage kahin use nahi karna.
```

```python
def add_print(a, b):
    print(a + b)

add_print(3,7)
```

```python
def add_return(a, b):
    return a + b     # value wapas bhej di, print nahi kiya

result = add_return(3, 7)    # result me value store ho gayi

if result > 5:
    print("Sum is greater:", result)
else:
    print("Sum is smaller:", result)
```
