# Histogram 

Simple ek line mein: Histogram batata hai ki tumhare data mein kaunsi values kitni baar aa rahi hain — matlab data ka "pattern" ya "shape" dikhata hai.

steps 
1. Data leta hai
2. Groups (bins) banata hai example 0-10, 10-20, 20-30
3. Counting karta hai
4. Bar banata hai

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Step 1: Load the dataset
df = pd.read_csv("student_marks_lpa.csv")
print(df.head())

# Step 2: Prepare X (input) and y (output)
X = df[["Marks"]]      # input feature -> 2D array chahiye sklearn ko
y = df["LPA"]           # output/target

# Step 3: Create and train the model
model = LinearRegression()
model.fit(X, y)

# Step 4: Get slope (m) and intercept (c) -> y = m*x + c
m = model.coef_[0]
c = model.intercept_
print(f"\nRegression Line: LPA = {m:.4f} * Marks + {c:.4f}")

# Step 5: Predict LPA for all marks (to draw the regression line)
y_pred = model.predict(X)

# Step 6: Predict for a new student (example: marks = 77)
new_marks = 77
predicted_lpa = model.predict([[new_marks]])
print(f"Predicted LPA for {new_marks} marks: {predicted_lpa[0]:.2f}")

# Step 7: Plot actual data points + regression line
plt.figure(figsize=(8,6))
plt.scatter(X, y, color="blue", label="Actual Data (Students)")
plt.plot(X, y_pred, color="red", linewidth=2, label="Regression Line")
plt.xlabel("Marks (out of 100)")
plt.ylabel("LPA (Package in Lakhs)")
plt.title("Linear Regression: Marks vs LPA")
plt.legend()
plt.grid(alpha=0.3)
plt.savefig("regression_plot.png", dpi=120)
plt.show()
```

## univariate analysis 

Graph	Best used for

Histogram	Distribution/frequency of a numerical variable (e.g., marks, height, age)

Box Plot	Spread, median, and outliers of a numerical variable

Bar Chart	Frequency/count of a categorical variable (e.g., number of students per grade)

Pie Chart	Proportion/percentage of categories within a whole

Density Plot (KDE)	Smoothed version of a distribution, shows shape more clearly than 

histogram

Count Plot	Count of occurrences for each category

## bivariate analysis 

Scatter Plot	Relationship between two numerical variables (e.g., Marks vs LPA — exactly 
what we did earlier!)

Line Plot	Trend of one numerical variable over another (often time), e.g., Sales over Months

Bar Chart (grouped/stacked)	Comparing a numerical variable across categories, e.g., 
Average LPA per Department

Box Plot (by category)	Comparing distribution of a numerical variable across different categories

Heatmap	Correlation between multiple numerical variables at once

Regression Plot	Scatter plot + best-fit line — shows relationship strength/direction (like our Marks vs LPA example)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Load dataset
df = pd.read_csv("student_marks_lpa.csv")

# Create Grade categories based on Marks (needed for some charts)
def get_grade(m):
    if m < 50: return "Low (<50)"
    elif m < 70: return "Medium (50-70)"
    elif m < 90: return "High (70-90)"
    else: return "Excellent (90+)"

df["Grade"] = df["Marks"].apply(get_grade)


# ============================================================
# 1. HISTOGRAM (Univariate) - Distribution of Marks
# ============================================================
plt.figure(figsize=(7,5))
plt.hist(df["Marks"], bins=6, color="skyblue", edgecolor="black")
plt.title("Histogram: Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.show()


# ============================================================
# 2. BOX PLOT (Univariate) - Spread of LPA
# ============================================================
plt.figure(figsize=(5,6))
plt.boxplot(df["LPA"], vert=True, patch_artist=True,
            boxprops=dict(facecolor="lightgreen"))
plt.title("Box Plot: Spread of LPA")
plt.ylabel("LPA (Lakhs)")
plt.show()


# ============================================================
# 3. BAR CHART (Univariate) - Number of Students per Grade Category
# ============================================================
grade_counts = df["Grade"].value_counts().reindex(
    ["Low (<50)", "Medium (50-70)", "High (70-90)", "Excellent (90+)"])
plt.figure(figsize=(7,5))
plt.bar(grade_counts.index, grade_counts.values, color="orange", edgecolor="black")
plt.title("Bar Chart: Number of Students per Grade Category")
plt.xlabel("Grade Category")
plt.ylabel("Number of Students")
plt.show()


# ============================================================
# 4. SCATTER PLOT (Bivariate) - Marks vs LPA
# ============================================================
plt.figure(figsize=(7,5))
plt.scatter(df["Marks"], df["LPA"], color="blue", s=60)
plt.title("Scatter Plot: Marks vs LPA")
plt.xlabel("Marks")
plt.ylabel("LPA (Lakhs)")
plt.grid(alpha=0.3)
plt.show()


# ============================================================
# 5. REGRESSION PLOT (Bivariate) - Scatter + Best Fit Line
# ============================================================
X = df[["Marks"]]
y = df["LPA"]
model = LinearRegression().fit(X, y)
y_pred = model.predict(X)

plt.figure(figsize=(7,5))
plt.scatter(X, y, color="blue", label="Actual Data")
plt.plot(X, y_pred, color="red", linewidth=2, label="Regression Line")
plt.title("Regression Plot: Marks vs LPA")
plt.xlabel("Marks")
plt.ylabel("LPA (Lakhs)")
plt.legend()
plt.grid(alpha=0.3)
plt.show()


# ============================================================
# 6. LINE PLOT (Bivariate) - LPA trend as Marks increase (sorted)
# ============================================================
df_sorted = df.sort_values("Marks")
plt.figure(figsize=(7,5))
plt.plot(df_sorted["Marks"], df_sorted["LPA"], marker="o", color="green")
plt.title("Line Plot: LPA Trend as Marks Increase")
plt.xlabel("Marks")
plt.ylabel("LPA (Lakhs)")
plt.grid(alpha=0.3)
plt.show()


# ============================================================
# 7. GROUPED BAR CHART (Bivariate) - Average LPA per Grade Category
# ============================================================
avg_lpa_per_grade = df.groupby("Grade")["LPA"].mean().reindex(
    ["Low (<50)", "Medium (50-70)", "High (70-90)", "Excellent (90+)"])
plt.figure(figsize=(7,5))
plt.bar(avg_lpa_per_grade.index, avg_lpa_per_grade.values, color="mediumpurple", edgecolor="black")
plt.title("Bar Chart: Average LPA per Grade Category")
plt.xlabel("Grade Category")
plt.ylabel("Average LPA (Lakhs)")
plt.show()


# ============================================================
# 8. BOX PLOT BY CATEGORY (Bivariate) - LPA distribution per Grade
# ============================================================
groups = [df[df["Grade"] == g]["LPA"].values for g in
          ["Low (<50)", "Medium (50-70)", "High (70-90)", "Excellent (90+)"]]
plt.figure(figsize=(7,5))
plt.boxplot(groups, tick_labels=["Low (<50)", "Medium (50-70)", "High (70-90)", "Excellent (90+)"],
            patch_artist=True, boxprops=dict(facecolor="lightblue"))
plt.title("Box Plot: LPA Distribution across Grade Categories")
plt.xlabel("Grade Category")
plt.ylabel("LPA (Lakhs)")
plt.show()


# ============================================================
# 9. PIE CHART (Univariate) - Proportion of Students per Grade
# ============================================================
plt.figure(figsize=(6,6))
plt.pie(grade_counts.values, labels=grade_counts.index, autopct="%1.1f%%", startangle=90)
plt.title("Pie Chart: Proportion of Students per Grade Category")
plt.show()


# ============================================================
# 10. HEATMAP (Correlation) - Marks vs LPA correlation
# ============================================================
corr = df[["Marks", "LPA"]].corr()
plt.figure(figsize=(5,4))
plt.imshow(corr, cmap="coolwarm", vmin=-1, vmax=1)
plt.xticks(range(len(corr.columns)), corr.columns)
plt.yticks(range(len(corr.columns)), corr.columns)
for i in range(len(corr.columns)):
    for j in range(len(corr.columns)):
        plt.text(j, i, f"{corr.iloc[i,j]:.2f}", ha="center", va="center", color="black")
plt.colorbar(label="Correlation")
plt.title("Heatmap: Correlation between Marks and LPA")
plt.show()
```
