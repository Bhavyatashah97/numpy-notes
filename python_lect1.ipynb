```python

# Integer (int)
age = 24
print(age)
print(type(age))

# Float
height = 5.7
print(height)
print(type(height))

# String (str)
name = "Bhavyata"
print(name)
print(type(name))

# Boolean (bool)
is_student = True
print(is_student)
print(type(is_student))

```

```python
| Data Type | Example          | Meaning                |
| --------- | ---------------- | ---------------------- |
| int       | `10`             | Whole number           |
| float     | `3.14`           | Decimal number         |
| str       | `"Hello"`        | Text                   |
| bool      | `True`, `False`  | Yes/No value           |
```
